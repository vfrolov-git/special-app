# Бинарный поиск
Функция binary_search получает отсортированный массив и значение. Если 
значение присутствует в массиве, то функция возвращает его позицию.П

```python
def binary_search(list, item):
    low = 0
    high = len(list) - 1

    while low <= high:
        mid = (low + high)
        #print(mid)
        guess = list[mid] # proposed item according position mid
        if guess == item:
            return mid   # item position
        if guess > item:
            high = mid - 1
        else:
            low = mid + 1

my_list = [1, 3, 5, 7, 8, 10]

print(binary_search(my_list, 10))
```

# Сортировка выбором 
*Smallest array element

```python
# Smallest array element
def find_samllest(arr):
    smallest = arr[0]
    smallest_index = 0
    for i in range(1, len(arr)):
        if arr[i] < smallest:
            smallest = arr[i]
            smallest_index = i
    return smallest_index
```

```python
# Selection sort
def selection_sort(arr):
    new_arr = []
    for i in range(len(arr)):
        smallest = find_samllest(arr)
        new_arr.append(arr.pop(smallest))
    return new_arr

print(selection_sort([5, 3, 6, 2, 12]))
```

# Быстрая сортировка
Опорный элемент, массив значений больше опорного и меньше опорного

```python
def quick_sort(my_array):
    if len(my_array) < 2:# массив без элемента или с одним уже отсортированы
        return my_array
    else:
        pivot = my_array[0]
        less = [i for i in my_array[1:] if i <= pivot] # подмассив всех меньше опорного
        greater = [i for i in my_array[1:] if i > pivot] # подмассив всех меньше опорного
        return quick_sort(less) + [pivot] + quick_sort(greater) # вызов функции внутри функции


test_array = [10, 5, 2, 7, 8, 1, 4, 4]
quick_sort(test_array)
```

```python
# Таблица стоимости узлов
costs = {}
costs['a'] = 6
costs['b'] = 2
costs['fin'] = infinity
costs
```

```python
# Код создания хеш-таблицы родителей
parents = {}
parents['a'] = 'start'
parents['b'] = 'start'
parents['fin'] = None
parents
```

```python
#  массив для отслеживания всех уже обработанных узлов, так как один узел
# не должен обрабатываться многократно
processed = []
```

```python
# функция find_lowest_cost_node узел с наименьшей стоимостью 
# Если это узел с наименьшей стоимостью из уже виденных 
# и он еще не был обработан...
def find_lowest_cost_node(costs): ## costs -> Таблица стоимости узлов
    lowest_cost = float('inf') # 'inf' - бесконечность
    lowest_cost_node = None
    for node in costs: # перебрать все узлы
        cost = costs[node]
        if cost < lowest_cost and node not in processed:
            lowest_cost = cost # ...он назначается новым узлом с наименьшей стоимостью
            lowest_cost_node = node
    return lowest_cost_node     
```

```python
node = find_lowest_cost_node(costs) # Ищем узел с наименьшей стоимостью с помощью функции (мс. выше)
while node is not None: # None-> final. Если обработаны все узлы, то цикл завершён.
    cost = costs[node]
    neighbors = graph_d[node]
    for n in neighbors.keys(): # Перебрать всех соседей текущего узла
        new_cost = cost + neighbors[n] # Если к соседу можно быстрее добраться ...
        if costs[n] > new_cost:        # ...через текущий узел...
            costs[n] = new_cost        # ...обновить стоимость для этого узла
            parents[n] = node          # Этот узел становится новым родителем для соседа
    processed.append(node)       # Узел помечается как обработанный
    node = find_lowest_cost_node(costs)
    

print(costs) # новая таблица стоимостей для быстрейшего перехода к fin (start -> b -> a -> fin)
```
